## 4. Definir pasta de saída

```python
import os
OUTPUT_DIR = "/content/drive/MyDrive/auto-video-producer/cache/omnivoice_outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"Áudios serão salvos em: {OUTPUT_DIR}")
```

## 5. Gerar áudio com OmniVoice

```python
from omnivoice import OmniVoice
import torch
import os

# Inicializa OmniVoice
model = OmniVoice(
    model_path=None,  # auto-download do HuggingFace
    device="cuda",  # use GPU T4
    compile=True,    # torch.compile para velocidade
    voice_preset="default"  # ou caminho para .npz de voice design
)

# Texto de exemplo
texto = """Bem-vindo ao Auto Video Producer!

Este é um exemplo de geração de áudio com OmniVoice.
O sistema converte texto em áudio com voice design,
perfeito para vídeos de produção profissional."""

# Gera áudio
audio_path = os.path.join(OUTPUT_DIR, "exemplo.wav")
model.generate(texto, audio_path, speed=1.0, temperature=0.7, top_p=0.9)
print(f"Áudio gerado: {audio_path}")
```

## 6. Ouvir o áudio gerado

```python
from IPython.display import Audio
Audio(audio_path)
```

## 7. (Opcional) Gerar áudio para um arquivo de texto

```python
# Exemplo: ler texto de um arquivo
texto_path = "/content/drive/MyDrive/auto-video-producer/input/textos/exemplo.txt"
if os.path.exists(texto_path):
    with open(texto_path, "r") as f:
        texto = f.read()
    audio_path = os.path.join(OUTPUT_DIR, "texto_personalizado.wav")
    model.generate(texto, audio_path, speed=1.0, temperature=0.7, top_p=0.9)
    print(f"Áudio gerado a partir de {texto_path}: {audio_path}")
    Audio(audio_path)
```
